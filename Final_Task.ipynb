```python
import pandas as pd
import numpy as np

# 1. Load dataset (Creating a mock dataset for this example)
data = {
    'Student_ID': [101, 102, 103, 104, 105],
    'Theory_Marks': [85, np.nan, 70, 95, 60],
    'Lab_Marks': [90, 80, 75, np.nan, 65]
}
df = pd.DataFrame(data)

# 2. Handle missing values
# Filling NaNs with the mean of their respective columns
df['Theory_Marks'] = df['Theory_Marks'].fillna(df['Theory_Marks'].mean())
df['Lab_Marks'] = df['Lab_Marks'].fillna(df['Lab_Marks'].mean())

# 3. Perform NumPy operations
# Converting columns to NumPy arrays for calculation
theory_array = df['Theory_Marks'].values
lab_array = df['Lab_Marks'].values

# 4. Broadcasting & arithmetic
# Calculate total marks (weighted: 60% Theory, 40% Lab)
total_marks = (theory_array * 0.6) + (lab_array * 0.4)
df['Total_Marks'] = total_marks

# 5. Feature engineering
# Assign Pass/Fail status
df['Status'] = np.where(df['Total_Marks'] >= 70, 'Pass', 'Fail')

# 6. Sorting & filtering
# Sort by highest marks and filter for students who passed
df_final = df[df['Status'] == 'Pass'].sort_values(by='Total_Marks', ascending=False)

# 7. Save final output
df_final.to_pdf('final_results.pdf', index=False)

# Use display() instead of print() to avoid your 'tuple' error
display(df_final)
```
